# Проверка загрузчика данных

Этот notebook проверяет, что `src.1_data_loader.loader` загружает `data/trade.xlsx` без предварительной обработки данных.

```python
from pathlib import Path
import sys

# Определяем корень проекта, даже если notebook запущен из папки tests_ipynb.
project_root = Path.cwd()
if not (project_root / "data" / "trade.xlsx").exists():
    project_root = project_root.parent

# Добавляем корень проекта в пути импорта, чтобы Python нашел модули src.
sys.path.insert(0, str(project_root))

data_path = project_root / "data" / "trade.xlsx"
data_path
```

```python
from importlib import import_module

loader = import_module("src.1_data_loader.loader")

# Загружаем файл как есть. Очистка будет выполняться в модулях preprocessing.
data = loader.load_excel_data(data_path)
data.head()
```

```python
# Базовые проверки загруженной таблицы.
expected_columns = ["Import_Export", "Year", "Month", "TNVED", "Country", "US_dollars"]

assert data_path.exists(), f"File not found: {data_path}"
assert list(data.columns) == expected_columns
assert len(data) > 0

print("Rows:", len(data))
print("Columns:", list(data.columns))
```

```python
# Проверяем, что загрузчик не заменял исходные русские значения.
print(data["Import_Export"].unique())
print(data.dtypes)
```
