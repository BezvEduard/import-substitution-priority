# Проверка preprocessing

Этот notebook проверяет подготовку данных после загрузки Excel-файла.

```python
from importlib import import_module
from pathlib import Path
import sys

# Определяем корень проекта, даже если notebook запущен из папки tests_ipynb.
project_root = Path.cwd()
if not (project_root / "data" / "trade.xlsx").exists():
    project_root = project_root.parent

# Добавляем корень проекта в пути импорта.
sys.path.insert(0, str(project_root))

loader = import_module("src.1_data_loader.loader")
preprocessing = import_module("src.2_preprocessing.preprocessing")
```

```python
# Загружаем данные и выполняем preprocessing.
raw_data = loader.load_excel_data(project_root / "data" / "trade.xlsx")
prepared_data = preprocessing.preprocess_trade_data(raw_data)

prepared_data.head()
```

```python
# Проверяем результат базовой подготовки.
assert prepared_data["Month"].between(1, 12).all()
assert set(prepared_data["Import_Export"].unique()) == {"Import", "Export"}
assert prepared_data["US_dollars"].gt(0).all()
assert "0" not in set(prepared_data["TNVED"])

prepared_data.info()
```

```python
# Формируем таблицы для будущего расчета критериев.
yearly_trade = preprocessing.make_yearly_trade_table(prepared_data)
country_import = preprocessing.make_country_import_table(prepared_data)

display(yearly_trade.head(30))
display(country_import.head(30))
```
